# Class Lab: Discover Association Rules by Apriori Algorithm

## Step 1: setup goal
- use Apriori algorithm to discover association rules

## Step 2: retrieve data

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
from pathlib import Path
source_path = Path.cwd() / '..' / 'csv' / 'market_dummy_data.csv'
df = pd.read_csv(source_path)
df
```

## Step 3: prepare data

```python
basket = pd.pivot_table(data=df, index='ID', columns='Item', values='Quantity', aggfunc='sum', fill_value=0)
basket
```

```python
def convert_into_binary(x): # 將購買數量轉換成0/1
    if x >= 1:
        return True
    else:
        return False

basket_set = basket.map(convert_into_binary)
basket_set
```

## Step 4: explore data

```python
# Find frequent itemsets
freq_itemsets = apriori(basket_set, min_support=0.4, use_colnames=True)
freq_itemsets.sort_values(by='support', ascending=False)
```

## Step 5: build model
Comparing the accuracy of the original data set with latent variables

```python
# Generate association rules
min_conf = 0.7
rules = association_rules(freq_itemsets, metric="confidence", min_threshold=min_conf)
rules[["antecedents", "consequents", "support", "confidence", "lift"]]
```
