```python
from pathlib import Path
import pandas as pd
```

# Load file

```python
source_file = Path.cwd() / '../csv/shoe_size.csv'
df = pd.read_csv(source_file)
df.columns = df.columns.str.replace('ID', 'id')
df.columns = df.columns.str.replace('Gender', 'gender')
df.columns = df.columns.str.replace('Height_cm', 'height_cm')
df.columns = df.columns.str.replace('Weight_kg', 'weight_kg')
df.columns = df.columns.str.replace('Shoe size_cm', 'shoe_size_cm')

df.head()
```

# Check and drop duplicates

```python
df.shape
```

```python
df[df.duplicated()] # 顯示重複的 rows
df[df.duplicated(keep=False)] # 顯示所有重複的 rows
df.drop_duplicates(inplace=True)
df.shape
```

# Check null values

```python
df.isnull().sum()
```

```python
# height_cm欄有5個null值，我們可以用其他的gender, shoe_size_cm 的平均值來補足height_cm的值
df[df['height_cm'].isnull()] # 顯示 height_cm 欄有空值的 rows
df[df['height_cm'].isnull()].index # row: 12, 53, 858, 899, 953)

df.groupby(['gender', 'weight_kg', 'shoe_size_cm'])['height_cm'].mean()

df['height_cm'] = df['height_cm'].fillna(value = df.groupby(['gender', 'weight_kg', 'shoe_size_cm'])['height_cm'].transform('mean'))

df.loc[[12, 53, 858, 899, 953], :]
# df[df['height_cm'].isnull()]
```

```python
# weight_kg欄有10個null值，我們可以用其他的gender, shoe_size_cm 的平均值來補足height_cm的值
df[df['weight_kg'].isnull()] # 顯示 height_cm 欄有空值的 rows
df[df['weight_kg'].isnull()].index # row: [202, 316, 350, 384, 432, 504, 695, 712, 730, 759]

df.groupby(['gender','height_cm', 'shoe_size_cm'])['weight_kg'].mean()

df['weight_kg'] = df['weight_kg'].fillna(value = df.groupby(['gender', 'height_cm', 'shoe_size_cm'])['weight_kg'].transform('mean'))

df.loc[[202, 316, 350, 384, 432, 504, 695, 712, 730, 759], :]
df[df['weight_kg'].isnull()]
```

```python
df[df['shoe_size_cm'].isnull()] # 顯示 shoe_size_cm 欄有空值的 rows
df.dropna(axis=0, inplace=True)
df.isnull().sum()
```

```python
# df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['gender'] = df['gender'].astype('int')
df['height_cm'] = df['height_cm'].astype('int')
df['weight_kg'] = df['weight_kg'].astype('int')
df['shoe_size_cm'] = df['shoe_size_cm'].astype('int')
# df['id'] = df['id'].astype('str')
df.head()
# df.set_index('id', inplace=True)

df.dtypes
```
