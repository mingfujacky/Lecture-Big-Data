# MAPPER

```python
from pathlib import Path

source_in_file_path = Path.cwd() /'..' / 'txt' / 'source_in.txt'
mapper_out_file_path = Path.cwd() /'..' / 'txt' / 'mapper_out.txt'
all_words = []
```

```python
# reading entire line from STDIN (standard input)

with open(source_in_file_path, 'r') as f_in:
    content = f_in.readlines()

for line in content: 
    line = line.strip().lower().replace('\t',' ').replace(',','').replace('.','')\
           .replace('"','').replace("'",'').replace(':','')\
           .replace('“','').replace('”','').replace('’','')
    words = line.split()
    all_words = all_words + words
print(all_words)
```

```python
# output each word with count 1
all_words.sort()   
with open(mapper_out_file_path, "w") as f_mapper_out:
    for word in all_words: 
        f_mapper_out.write(f'{word}\t1\n')
```

# REDUCER

```python
#Reducer.py
from operator import itemgetter 
import sys 
from pathlib import Path

mapper_in_file_path =  Path.cwd() /'..' / 'txt' / 'mapper_out.txt'
reducer_out_file_path = Path.cwd() /'..' / 'txt' / 'reducer_out.txt'

with open(mapper_in_file_path, 'r') as f_in:
    content = f_in.readlines()
print(type(content))    
print(content)
```

```python
current_word = None
current_count = 0

with open(reducer_out_file_path, "w") as f_reducer_out:
# read the element from list one by one
    for element in content:
        if element!='':
            word, count = element.strip().split('\t')
            count = int(count) 
        # this IF-switch only works because Hadoop SORTS map output by key (word) before passed to reducer
        if current_word == word:
            current_count += count
        else: 
            # write result to STDOUT
            if current_word: 
                f_reducer_out.write(f'{current_word}\t{current_count}\n')
            current_word = word
            current_count = count

    # do not forget to output the last word if needed! 
    if current_word == word:
        f_reducer_out.write(f'{current_word}\t{current_count}\n')
```
