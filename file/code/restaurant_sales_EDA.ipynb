-	We explored:

-	Distributions

-	Time trends

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../csv/restaurant_sales.csv")
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

## 4. Univariate Analysis

```python
# Category counts
df['Category'].value_counts().plot(kind='bar')
plt.title("Item Categories")
plt.show()
```

# Bivariate Analysis

```python
sns.boxplot(x='Category', y='Price', data=df)
plt.title("Price per Category")
plt.show()
```

# Time Series

```python
df.groupby('Date')['Price'].sum().plot(kind='line', marker='o')
plt.title("Daily Sales Trend")
plt.show()
```

# 7. Customer Insights

```python
print("Unique Customers:", df['CustomerID'].nunique())
df.groupby('CustomerID')['Price'].sum().sort_values()
```
