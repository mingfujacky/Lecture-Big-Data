# Class Lab: Grouping similar observations to gain insight

## Step 1: setup goal
- uses an iris (鳶尾花) data set to group them into different types of irises.
- use k-means to divide data set into subsets (clusters), wherein observations
should be similar to those in the same cluster but differ greatly from the observations in other clusters.

## Step 2: retrieve data
- Scikit-learn provides a built-in dataset for iris data set, often used as a benchmark for clustering algorithms.
- This dataset, accessible via load_iris(), contains 3 types of iris.
  - Data instances: 150 (50 in each type)
  - Features
    - sepal(萼片) length in cm
    - sepal(萼片) width in cm
    - petal(花瓣) length in cm
    - petal(花瓣) width in cm
  - Target: 0-Setosa(山鳶尾), 1-Versicolour(變色鳶尾), 2-Virginica(維吉尼亞鳶尾)

```python
# import libraries
import numpy as np
from matplotlib import pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_iris
import pandas as pd
```

```python
# 1. 載入 Iris 真實數據集
iris = load_iris()
# 為了視覺化，我們只選取前兩個特徵：花萼長度 (Sepal Length) 和 花萼寬度 (Sepal Width)
X = iris.data[:, :2] 
feature_names = iris.feature_names[:2]
```

```python
# 2. 應用 K-means 演算法
# Iris 數據集有 3 個類別，所以我們設定 K=3
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
# 獲取聚類標籤和中心點
labels = kmeans.labels_
centers = kmeans.cluster_centers_
print("Cluster Labels:\n", labels)
print("Cluster Centers:\n", centers)
```

```python
# 3. 視覺化結果
plt.figure(figsize=(10, 7))

# 繪製所有資料點，並根據聚類結果上色
scatter = plt.scatter(X[:, 0], X[:, 1], c=labels)

# 繪製聚類中心點
plt.scatter(centers[:, 0], centers[:, 1], color='red', marker='X', label='Cluster Centers')

# 添加圖例
plt.legend()

plt.title(f'K-means Clustering on Iris Dataset (K=3)')
plt.xlabel(feature_names[0])
plt.ylabel(feature_names[1])
```
