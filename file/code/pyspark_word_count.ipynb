```python
# Word count application by PySpark / RDD
import re
from pathlib import Path
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("pyspark_word_count").getOrCreate()
txt_in_file_path = Path.cwd() /'..' / 'txt' / 'source_in.txt'

# Read the input file into rdd
rdd = spark.sparkContext.textFile(txt_in_file_path.as_posix())
```

```python
# 定義清理函數
def clean_text(text):
    text = text.lower()  # 轉小寫
    text = re.sub(r'[^\w\s]', '', text)  # 去除標點符號
    text = re.sub(r'\s+', ' ', text).strip()  # 去除多餘空白
    return text

# 清理，分詞，並計算詞頻
word_clean = rdd.map(clean_text)  # 清理文本
word_individual = word_clean.flatMap(lambda line: line.split()) # 分詞
word_pair = word_individual.map(lambda word: (word, 1)) # 記錄詞頻
word_count = word_pair.reduceByKey(lambda a, b: a + b) # 合併相同的詞
# word_counts = (rdd
#                .map(clean_text)  # 清理文本
#                .flatMap(lambda line: line.split())  # 分詞
#                .map(lambda word: (word, 1))  # 記錄詞頻
#                .reduceByKey(lambda a, b: a + b))  # 合併相同的詞

# 保存結果
rdd_out_file_path = Path.cwd() /'..' / 'txt' / 'rdd_out'
if rdd_out_file_path.exists():
    raise FileExistsError(f"{rdd_out_file_path.as_posix()} already exists. Please remove it before running the script.")

word_count.saveAsTextFile(rdd_out_file_path.as_posix())
```

```python
# read data from "../txt/rdd_out"
rdd_output = spark.sparkContext.textFile(rdd_out_file_path.as_posix())
for line in rdd_output.collect():
    print(line)
```
