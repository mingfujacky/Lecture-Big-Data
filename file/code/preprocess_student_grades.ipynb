### Clean data

```python
# Remove rows with any -1 scores 缺考以-1表示
filter = (df['exam1'] == -1) | (df['exam2'] == -1) | (df['exam3'] == -1) | (df['exam4'] == -1) | (df['exam5'] == -1)
i = df[filter].index
print(i)
df.loc[i]
df.drop(i, axis = 0, inplace = True)
```

```python
# Handle missing values
df.isnull() # True:空值 False:非空值
df.isnull().sum() # 每個欄位有多少空值數
```

```python
# Fill missing values with the mean of each exam
df[df['exam2'].isnull()] # 顯示exam2欄有空值的rows
df['exam2'] = df['exam2'].fillna(value = df['exam2'].mean())
df['exam3'] = df['exam3'].fillna(value = df['exam3'].mean())
df['exam5'] = df['exam5'].fillna(value = df['exam5'].mean())
df.isnull().sum() # 確認沒有空值了
df.describe()
```

### Transfer data

```python
# In sex column, change '男' to 1 and '女' to 0
df['sex'] = df['sex'].map({'男': 1, '女': 0})
df['sex'] = df['sex'].astype('int8')
df.head()
```

# Pack script of data preprocessing

```python
from pathlib import Path
import pandas as pd

# Load data
source_file = Path.cwd() / '../csv/student_scores.csv'
df = pd.read_csv(source_file)

# Change column names to english use replace method
df = df.rename(columns={
    'ID': 'id',
    '第1次平時考': 'exam1',
    '第2次平時考': 'exam2',
    '第3次平時考': 'exam3',
    '第4次平時考': 'exam4',
    '第5次平時考': 'exam5'
})

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Remove rows with any -1 scores 缺考以-1表示
filter = (df['exam1'] == -1) | (df['exam2'] == -1) | (df['exam3'] == -1) | (df['exam4'] == -1) | (df['exam5'] == -1)
i = df[filter].index
df.drop(i, axis = 0, inplace = True)

# Handle missing values
df['exam2'] = df['exam2'].fillna(value = df['exam2'].mean())
df['exam3'] = df['exam3'].fillna(value = df['exam3'].mean())
df['exam5'] = df['exam5'].fillna(value = df['exam5'].mean())

# In sex column, change '男' to 1 and '女' to 0
df['sex'] = df['sex'].map({'男': 1, '女': 0})

# Change data types
df['id'] = df['id'].astype('str')
df['exam1'] = df['exam1'].astype('int8')
df['exam2'] = df['exam2'].astype('int8')
df['exam3'] = df['exam3'].astype('int8')
df['exam4'] = df['exam4'].astype('int8')
df['exam5'] = df['exam5'].astype('int8')
df['sex'] = df['sex'].astype('int8')

# Reindex the DataFrame
df.reset_index(drop=True, inplace=True)

# Save cleaned data to new CSV file
df.head()
df.info()
output_file = Path.cwd() / '../csv/student_scores_cleaned.csv'
df.to_csv(output_file, index=False)
```
