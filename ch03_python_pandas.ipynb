# Operate Series
- Series 是一維陣列，類似Excel中的一欄
- Series 可以內定索引(0, 1, 2, ...)或自訂索引，用來標示資料的位置，類似於Excel中的列號
- Series 可以進行各種資料操作和分析，例如篩選、排序、統計等
- Series 的建立語法: `pd.Series(data, index=index)`

```python
# Create Series
import pandas as pd
name_list = ['王小明','李小美','陳大同','林小玉']
score_list = [65, 90, 81, 79]
# scores = pd.Series(score_list)
scores = pd.Series(score_list, index=name_list)

scores
```

```python
# Understand Series statistics
scores.describe() # summary statistics
```

```python
# Indexing Series
print(scores['王小明']) # access by index label
print(scores.iloc[0]) # access by index position
print(scores.loc['王小明']) # access by index label
# print(scores[1])     # access by index position (not recommended)
# print(scores.ENG)    # access by attribute (not recommended)
```

```python
# Modify Series
scores['王小明'] = 100 # modify by index label
scores.iloc[2] = 50     # modify by index position
scores.loc['林小玉'] = 95    # modify by index label
scores
```

```python
# Slicing Series
scores['王小明':'陳大同'] # 連續 slicing by index label
scores[1:3] # 連續 slicing by index position (not include the end position)
scores[['陳大同','王小明', '林小玉']] # 不連續 slicing
```

# Operate DataFrame
- DataFrame 是一個二維的資料結構，類似於Excel中的試算表
- DataFrame 由多個 Series 組成，每個 Series 是 DataFrame 的一欄
- DataFrame 可以有列號和欄號的內定索引(0, 1, 2, ...)或自訂列號和欄號索引，用來標示資料的位置
- DataFrame 可以進行各種資料操作和分析，例如篩選、排序、分組、合併等
- DataFrame 的建立語法: `pd.DataFrame(data, index=index, columns=columns)`

```python
# Create DataFrame
import pandas as pd
# 設定row index 跟column label
name_list = ['王小明','李小美','陳大同','林小玉']
subject_list = ['國文','英文','數學','自然','社會']
credit_list=  [[65,92,78,83,70],
       [90,72,76,93,56],
       [81,85,91,89,77],
       [79,53,47,94,80],
       ]
df = pd.DataFrame(credit_list,
                  index=name_list,
                  columns=subject_list)
df
```

```python
# Understand DataFrame
df.shape
# df.dtypes
# df.describe()
# df.head(3)
# df.tail(2)
```

```python
# Read single row and column from DataFrame
# df['英文'] # access by column label, return a Series
# df.loc['王小明'] # access by row index label, return a Series
# df.iloc[0]      # access by row index position, return a Series
print(df.loc['王小明','英文']) # access by row index label and column label
print(df.iloc[0,1])       # access by row index position and column position
df
```

```python
# Read multiple rows and columns from DataFrame
df[['英文','自然']]
df.loc[['王小明','林小玉']]
df.iloc[[1,2]]
df.loc[['王小明','林小玉'],['英文','自然']]
df.loc[:,'英文':'自然']
df.iloc[0:2,1:4]
df.iloc[[0,3],[1,3]]
```
